## แบบฟอร์ม Insight Card

ทุก insight เขียนในรูปแบบนี้ 5 บรรทัด ไม่ต้องยาว

## ตอบ 4 คำถามหลัก

1. **คนโพสต์เมื่อไร:** จำนวนโพสต์ ช่วงเวลาที่ข้อมูลครอบคลุม และชั่วโมงที่โพสต์มากที่สุด
2. **ใช้ภาษาอะไร:** รายงานภาษาหลักและสัดส่วน เพื่อบอกว่าถ้าทำคอนเทนต์ควรรองรับภาษาใด
3. **คนพูดถึงอะไร:** หา hashtag หรือคำ/วลีเด่นที่น่าสนใจ และยกตัวอย่างข้อความจริงประกอบ
4. **ข้อมูลนี้บอกอะไรได้และบอกอะไรไม่ได้:** สร้าง visualization 1 รูป แล้วเขียนข้อจำกัด 3–5 บรรทัด

> Part 1 เป็น warm-up ไม่ต้องทำ influencer analysis หรือ reply-network เว้นแต่สนใจทำเพิ่มเอง


```python
# ----------------- Your code here -----------------
# ใบ้: อ่าน JSON lines ทีละบรรทัดด้วย generator ไม่ต้องโหลดทั้งไฟล์
# def iter_jsonl(path):
#     with open(path, encoding='utf-8') as f:
#         for line in f:
#             yield json.loads(line)
def iter_jsonl(path):
    with open(path, encoding='utf-8') as f:
        for line in f:
            yield json.loads(line)

file1 = "/content/posts_20241127_075630.jsonl"
file2 = "/content/posts_20241127_082640.jsonl"

rows = []

for row in iter_jsonl(file1):
    rows.append(row)

for row in iter_jsonl(file2):
    rows.append(row)

df = pd.DataFrame(rows)


display(df.head())
```

```python
# 1
df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce", utc=True)

valid_time = df["created_at"].dropna()
if valid_time.empty:
    raise ValueError("ไม่พบค่า created_at ที่แปลงเป็นวันเวลาได้")

df["hour_utc"] = df["created_at"].dt.hour
hour_count = df["hour_utc"].value_counts().sort_index()

start_time = valid_time.min()
end_time = valid_time.max()
peak_hour = int(hour_count.idxmax())
peak_count = int(hour_count.max())

print("จำนวนโพสต์ทั้งหมด:", f"{len(df):,}", "โพสต์")
print("ช่วงเวลาเริ่มต้น:", start_time)
print("ช่วงเวลาสิ้นสุด:", end_time)
print(f"ชั่วโมงที่มีโพสต์มากที่สุด (UTC): {peak_hour:02d}:00–{peak_hour:02d}:59")
print("จำนวนโพสต์ในชั่วโมงนั้น:", f"{peak_count:,}", "โพสต์")

print("\nตีความ:")
print(
    f"ข้อมูลชุดนี้ครอบคลุมตั้งแต่ {start_time} ถึง {end_time} "
    f"และช่วง {peak_hour:02d}:00–{peak_hour:02d}:59 UTC มีการโพสต์มากที่สุด "
    "จึงเป็นช่วงเวลาที่เห็น activity สูงที่สุดในตัวอย่างข้อมูลที่นำมาวิเคราะห์"
)
```

```python
# ===== Q2: ใช้ภาษาอะไร =====
if "predicted_language" in df.columns:
    lang = (
        df["predicted_language"]
        .astype("string")
        .str.strip()
        .replace({"": pd.NA, "None": pd.NA, "nan": pd.NA})
    )
    language_source = "predicted_language"
else:
    # fallback กรณีไฟล์ไม่มี predicted_language
    !pip install -q langdetect
    from langdetect import detect, DetectorFactory
    DetectorFactory.seed = 42

    def detect_language(text):
        try:
            text = str(text).strip()
            return detect(text) if len(text) >= 3 else None
        except:
            return None

    lang = df["text"].apply(detect_language)
    language_source = "langdetect"

lang_count = lang.value_counts(dropna=True)
lang_percent = lang.value_counts(normalize=True, dropna=True).mul(100)

lang_table = pd.DataFrame({
    "จำนวนโพสต์": lang_count,
    "สัดส่วน (%)": lang_percent.round(2)
})

print("แหล่งข้อมูลภาษา:", language_source)
display(lang_table.head(10))

top_lang = lang_table.index[0]
top_lang_pct = float(lang_table.iloc[0]["สัดส่วน (%)"])

print(
    f"\nตีความ: ภาษาที่พบมากที่สุดคือ '{top_lang}' คิดเป็นประมาณ {top_lang_pct:.2f}% "
    "ของโพสต์ที่ระบุภาษาได้ ดังนั้นถ้าจะทำคอนเทนต์จากกลุ่มข้อมูลนี้ "
    f"ควรรองรับภาษา '{top_lang}' เป็นหลัก และพิจารณาภาษาอันดับถัดไปเพิ่มเติมหากมีสัดส่วนมากพอ"
)

ax = lang_table.head(10)["สัดส่วน (%)"].plot(kind="bar", figsize=(10, 5))
ax.set_title("Top 10 Languages in Bluesky Sample")
ax.set_xlabel("Language")
ax.set_ylabel("Percentage of Posts")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# ===== Q3: คนพูดถึงอะไร =====
text_series = df["text"].fillna("").astype(str)

# ดึง hashtag รองรับตัวอักษร/ตัวเลข/underscore หลัง #
all_hashtags = []
for text in text_series:
    all_hashtags.extend(re.findall(r"(?<!\w)#\w+", text.lower(), flags=re.UNICODE))

hashtag_count = Counter(all_hashtags)
top_hashtags = hashtag_count.most_common(10)

print("Top 10 hashtags:")
if top_hashtags:
    display(pd.DataFrame(top_hashtags, columns=["hashtag", "จำนวนครั้ง"]))
else:
    print("ไม่พบ hashtag ในตัวอย่างข้อมูลนี้")

# หา keyword เบื้องต้นเพิ่มเติม โดยตัด URL / mention / hashtag / ตัวเลขล้วน
basic_stopwords = {
    "the","a","an","and","or","to","of","in","on","for","is","it","this","that",
    "i","you","we","they","he","she","be","are","was","were","with","as","at",
    "my","your","our","their","but","not","from","by","have","has","had","do",
    "does","did","so","if","just","about","can","will","would","me","us"
}

tokens = []
for text in text_series.str.lower():
    text = re.sub(r"https?://\S+|www\.\S+|@\w+|#\w+", " ", text)
    words = re.findall(r"[a-zA-Z]{3,}", text)
    tokens.extend(w for w in words if w not in basic_stopwords)

top_words = Counter(tokens).most_common(15)
print("\nคำภาษาอังกฤษที่พบเด่น (หลังตัด stopwords แบบพื้นฐาน):")
display(pd.DataFrame(top_words, columns=["word", "จำนวนครั้ง"]).head(15))

# ยกตัวอย่างข้อความจริงจาก hashtag อันดับ 1
if top_hashtags:
    top_tag = top_hashtags[0][0]
    mask = text_series.str.lower().str.contains(
        rf"(?<!\w){re.escape(top_tag)}(?!\w)", regex=True, na=False
    )
    examples = df.loc[mask, ["text"]].dropna().drop_duplicates().head(3)

    print(f"\nตัวอย่างข้อความจริงที่มี {top_tag}:")
    for i, text in enumerate(examples["text"], 1):
        clean = re.sub(r"\s+", " ", str(text)).strip()
        print(f"{i}. {clean[:500]}")

    print(
        f"\nตีความ: hashtag ที่พบมากที่สุดคือ {top_tag} จำนวน {top_hashtags[0][1]:,} ครั้ง "
        "แต่ความถี่อย่างเดียวไม่เพียงพอที่จะสรุปว่าเป็นประเด็นสำคัญที่สุด "
        "จึงต้องดูข้อความตัวอย่างประกอบเพื่อเข้าใจบริบทด้วย"
    )
else:
    print(
        "\nตีความ: ตัวอย่างข้อมูลนี้ไม่พบ hashtag เพียงพอ จึงควรใช้คำ/วลีเด่น "
        "หรือเทคนิค n-gram แทนในการสรุปว่าคนกำลังพูดถึงอะไร"
    )
```

```python
# ===== Q4: Visualization + ข้อจำกัด =====
hour_plot = hour_count.reindex(range(24), fill_value=0)

plt.figure(figsize=(11, 5))
plt.bar(hour_plot.index, hour_plot.values)
plt.xticks(range(24))
plt.xlabel("Hour of day (UTC)")
plt.ylabel("Number of posts")
plt.title("Bluesky Posting Activity by Hour (UTC)")
plt.tight_layout()
plt.show()

print("ข้อจำกัดของข้อมูล")
print("1) ข้อมูลนี้มาจากโพสต์สาธารณะใน Bluesky และใช้เพียง 1–3 ไฟล์ จึงไม่ใช่ตัวแทนของผู้ใช้ Bluesky ทั้งหมด")
print("2) ช่วงเวลาที่พบมากที่สุดเป็นเวลา UTC และอาจไม่ตรงกับเวลาท้องถิ่นของผู้โพสต์แต่ละคน")
print("3) ภาษาใน predicted_language เป็นผลจากการทำนาย จึงอาจจำแนกผิดได้ โดยเฉพาะข้อความสั้นหรือข้อความที่ใช้หลายภาษา")
print("4) Hashtag/คำที่พบถี่บอกได้ว่าอะไรถูกพูดถึงบ่อย แต่ยังบอกไม่ได้ว่าผู้โพสต์รู้สึกบวกหรือลบต่อเรื่องนั้น")
print("5) การวิเคราะห์นี้อธิบายพฤติกรรมในตัวอย่างข้อมูลช่วงเวลานี้เท่านั้น ไม่ควรใช้สรุปแนวโน้มระยะยาวโดยไม่มีข้อมูลเพิ่ม")
```
